```python
# Se você criar uma variável com o mesmo nome dentro de uma função, 
# essa variável será local e só poderá ser usada dentro da função. 
# A variável global com o mesmo nome permanecerá como estava, global e com o valor original.

x = "incrível"

def minha_função2():
  x = "fantástico"
  print("Python é " + x) # isso vai fazer ela mostra o resutado

minha_função2() # Chamou -> def minha_função2(): x = "fantástico" print("Python é " + x)
 
print("Python é " + x)
```

```python
# Se você usar a globalpalavra-chave, a variável pertencerá ao escopo global:

x = "incrível"  # Para alterar o valor de uma variável global dentro de uma função, 
                # faça referência à variável usando a globalpalavra-chave:

def minha_função3():
  global x
  x = "fantástico"

minha_função3() # usa esta função para pode retorna  o "X" global

print("Python é " + x) # para mostra o resutado precisa fazer o "print"
```

```python
# Variáveis ​​de tipo numérico são criadas quando você atribui um valor a elas:

# Int, ou inteiro, é um número inteiro, positivo ou negativo, sem decimais, de comprimento ilimitado.
x = 1 # int
y = 35656222554887711 # int
z = -3255522 # int

# Para verificar o tipo de qualquer objeto em Python, use a type()função:
print(type(x))
print(type(y))
print(type(z))

# Um número de ponto flutuante, ou "float number", é um número, positivo ou negativo, que contém uma ou mais casas decimais.
x = 1.10 #float
y = 1.0 #float
z = -35.59 #float

# Números de ponto flutuante também podem ser números científicos com um "e" para indicar a potência de 10.
a = 35e3 # float
b = 12E4 # float
c = -87.7e100 # float

# Para verificar o tipo de qualquer objeto em Python, use a type()função:
print(type(x))
print(type(y))
print(type(z))
print(type(a))
print(type(b))
print(type(b))

# Os números complexos são escritos com um "j" representando a parte imaginária:
x = 3+5j # complex
y = 5j # complex
z = -5j # complex

# Para verificar o tipo de qualquer objeto em Python, use a type()função:
print(type(x))
print(type(y))
print(type(z))

# Conversão de tipo
# ocê pode converter de um tipo para outro com os métodos int()`convert` float(), `convert` e `convert` complex():
x = 1    # int
y = 2.8  # float
z = 1j   # complex

#convert from int to float:
a = float(x)

#convert from float to int:
b = int(y)

#convert from int to complex:
c = complex(x)

print(a)
print(b)
print(c)

print(type(a))
print(type(b))
print(type(c))
```

```python
# String de verificação
# Para verificar se uma determinada frase ou caractere está presente em uma string, podemos usar a palavra-chave in.
txt = "As melhores coisas da vida são de graça!"
print("de graça" in txt) # Verifique se a palavra "graça" está presente no texto a seguir:

# Imprimir somente se "grátis" estiver presente
txt = "As melhores coisas da vida são de graça!"
if "de graça" in txt:
  print("Sim, 'de graça' está presente.")

# Para verificar se uma determinada frase ou caractere NÃO está presente em uma string, podemos usar a palavra-chave not in
txt = "As melhores coisas da vida são de graça!"
print("caro" not in txt) # Verifique se a palavra "caro" NÃO está presente no texto a seguir:

# Imprimir somente se "caro" NÃO estiver presente:
txt = "As melhores coisas da vida são de graça!"
if "caro" not in txt:
  print("Não, 'caro' NÃO está presente.")
```

```python
# Você pode retornar um intervalo de caracteres usando a sintaxe de fatiamento.
b = "Oláa, Mundo!"
print(b[2:5])

# Corte desde o início
b = "Oláa, Mundo!"
print(b[:5]) # Obtenha os caracteres do início até a posição 5 (não incluídos)

# Corte até o final
b = "Oláa, Mundo!"
print(b[2:]) # Obtenha os caracteres da posição 2 até o final

# Indexação negativa
b = "Oláa, Mundo!"
print(b[-5:-2]) # Use índices negativos para iniciar o recorte a partir do final da string
```

```python
# O upper()método retorna a string em maiúsculas:
a = "Oláa, Mundo!"
print(a.upper())

# O lower()método retorna a string em letras minúsculas:
a = "Oláa, Mundo!!"
print(a.lower())

# Espaço em branco é o espaço antes e/ou depois do texto propriamente dito, e muitas vezes você deseja remover esse espaço.
a = "   Oláa, Mundo!!!  "
print(a.strip()) # retorna "Oláa, Mundo!!!"

# O replace()método substitui uma string por outra string
a = "Oláa, Mundo!!!!"
print(a.replace("O", "J"))
print(a.replace("a", "i"))

# O split()método divide a string em substrings se encontrar ocorrências do separador:
a = "Oláa, Mundo!!!!!"
print(a.split(",")) # returns ['Oláa', ' Mundo!!!!!']
```

```python
# Um marcador de posição pode incluir um modificador para formatar o valor.

# Um modificador é incluído adicionando dois pontos : seguidos por um tipo de formatação válido, como , .2f que significa número de ponto fixo com 2 casas decimais:
preco = 59
txt = f"O preço é R$ {preco:.2f} Reais" # Exiba o preço com 2 casas decimais
print(txt)

# Realize uma operação matemática no espaço reservado e retorne o resultado
txt = f"O preço é R$ {20 * 59} Reais"
print(txt)
```

```python
# O caractere de escape permite usar aspas duplas em situações onde normalmente não seria permitido

# \'	Aspas simples
txt = "1. Nós somos os chamados \'Vikings\' do norte."
print(txt) 

# \\    Barra invertida
txt = "2.Isso inserirá uma \\ (barra invertida)."
print(txt)

# \n    Nova linha
txt = "3. Olá\nMundo!"
print(txt)

# \r    Carregar Retornar
txt = "Olá \r4. Mundo!" # Cade 'Olá'
print(txt)

# \t    Tab
txt = "5. Olá \tMundo!" 
print(txt)

# \b	Backspace
txt = "6. Olá \bMundo!" # apaga um caractere (backspace)
print(txt)

# \ooo  Octal value
# #Uma barra invertida seguida por três números inteiros resultará em um valor octal:
txt = "7. \157\154\141\040\155\165\156\144\157" # representação em código octal (ASCII)
print(txt) 

# \xhh	Hex value
# Uma barra invertida seguida de um 'x' e um número hexadecimal 
txt = "8. \x6f\x6c\x61\x20\x6d\x75\x6e\x64\x6f" #representa um valor hexadecimal
print(txt) 
```

```python
# Métodos de string
# Conjunto de métodos integrados que você pode usar em strings.

exemplo_de_variável = "A string que vai ser alterada"

exemplo_de_variável.capitalize()	# Converte o primeiro caractere para maiúsculo
exemplo_de_variável.casefold()  	# Converte a string para minúsculo
exemplo_de_variável.center()	    # Retorna uma string centralizada
exemplo_de_variável.count()     	# Retorna o número de vezes que um valor especificado ocorre em uma string
exemplo_de_variável.encode()	    # Retorna uma versão codificada da string
exemplo_de_variável.endswith()	    # Retorna True se a string terminar com o valor especificado
exemplo_de_variável.expandtabs()	# Define o tamanho da tabulação da string
exemplo_de_variável.find()          # Busca um valor especificado na string e retorna a posição onde ele foi encontrado
exemplo_de_variável.format()	    # Formata valores especificados em uma string
exemplo_de_variável.format_map()	# Formata valores especificados em uma string
exemplo_de_variável.index()	        # Busca um valor especificado na string e retorna a posição onde ele foi encontrado
exemplo_de_variável.isalnum()	    # Retorna True se todos os caracteres da string forem alfanuméricos
exemplo_de_variável.isalpha()	    # Retorna True se todos os caracteres da string pertencerem ao alfabeto
exemplo_de_variável.isascii()	    # Retorna True se todos os caracteres da string forem caracteres ASCII
exemplo_de_variável.isdecimal() 	# Retorna True se todos os caracteres da string forem decimais
exemplo_de_variável.isdigit()	    # Retorna True se todos os caracteres da string forem dígitos
exemplo_de_variável.isidentifier()	# Retorna True se a string for um identificador
exemplo_de_variável.islower()	    # Retorna True se todos os caracteres da string estiverem em minúsculo
exemplo_de_variável.isnumeric() 	# Retorna True se todos os caracteres da string forem numéricos
exemplo_de_variável.isprintable()	# Retorna True se todos os caracteres da string forem imprimíveis
exemplo_de_variável.isspace()	    # Retorna True se todos os caracteres da string forem espaços em branco
exemplo_de_variável.istitle()	    # Retorna True se a string seguir as regras de um título
exemplo_de_variável.isupper()	    # Retorna True se todos os caracteres da string estiverem em letras maiúsculas
exemplo_de_variável.join()          # Une os elementos de um iterável ao final da string
exemplo_de_variável.ljust() 	    # Retorna uma versão da string alinhada à esquerda
exemplo_de_variável.lower()	        # Converte uma string para letras minúsculas
exemplo_de_variável.lstrip()	    # Retorna uma versão da string sem espaços à esquerda
exemplo_de_variável.maketrans()	    # Retorna uma tabela de tradução para ser usada em traduções
exemplo_de_variável.partition()	    # Retorna uma tupla na qual a string é dividida em três partes
exemplo_de_variável.replace()   	# Retorna uma string onde um valor especificado é substituído por outro valor especificado
exemplo_de_variável.rfind()     	# Busca um valor especificado na string e retorna a última posição onde ele foi encontrado
exemplo_de_variável.rindex()    	# Busca um valor especificado na string e retorna a última posição onde ele foi encontrado
exemplo_de_variável.rjust()     	# Retorna uma versão da string alinhada à direita
exemplo_de_variável.rpartition()	# Retorna uma tupla na qual a string é dividida em três partes
exemplo_de_variável.rsplit()	    # Divide a string no separador especificado e retorna uma lista
exemplo_de_variável.rstrip()	    # Retorna uma versão da string sem espaços à direita
exemplo_de_variável.split()         # Divide a string no separador especificado e retorna uma lista
exemplo_de_variável.splitlines()	# Divide a string nas quebras de linha e retorna uma lista
exemplo_de_variável.startswith()	# Retorna True se a string começar com o valor especificado
exemplo_de_variável.strip()	        # Retorna uma versão da string sem espaços nas extremidades
exemplo_de_variável.swapcase()	    # Inverte as maiúsculas e minúsculas; minúsculas viram maiúsculas e vice-versa
exemplo_de_variável.title()	        # Converte o primeiro caractere de cada palavra para maiúsculo
exemplo_de_variável.translate()     # Retorna uma string traduzida
exemplo_de_variável.upper()	        # Converte uma string para letras maiúsculas
exemplo_de_variável.zfill()         # Preenche a string com uma quantidade especificada de zeros no início

# Observação: Todos os métodos de string retornam novos valores. Eles não alteram a string original
```
