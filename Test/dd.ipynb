```python
import requests

def obter_sas_token():
    # Definição do host e endpoint conforme o apicipv1.json
    # IMPORTANTE: Substitua 'Caminho_backend' pelo domínio real do seu servidor
    base_url = "https://Caminho_backend/api/cip/v1"
    endpoint = f"{base_url}/sas-tokens"
    
    # Cabeçalhos da requisição
    headers = {
        "Accept": "application/json"
        # Se a sua API precisar de um token de autorização, descomente e adicione a linha abaixo:
        # "Authorization": "Bearer SEU_TOKEN_AQUI"
    }

    try:
        # Realizando a requisição GET
        response = requests.get(endpoint, headers=headers)
        
        # O método raise_for_status() levanta uma exceção para códigos de erro (4xx ou 5xx)
        response.raise_for_status()
        
        # Convertendo a resposta JSON para um dicionário Python
        resposta_json = response.json()
        
        # Extraindo a URI navegando pela estrutura: { "data": { "sasToken": "..." }, "links": {...} }
        sas_token = resposta_json.get("data", {}).get("sasToken")
        
        # Acessando o header de resposta 'x-v' descrito na documentação
        versao_api = response.headers.get("x-v", "Versão não informada")
        
        print(f"Versão da API retornada no header: {versao_api}")
        print(f"SAS Token (URI): {sas_token}")
        
        return sas_token

    except requests.exceptions.HTTPError as http_err:
        # Tratamento de erros detalhados mapeados na documentação
        print(f"Falha na requisição. Código: {response.status_code}")
        print(f"Detalhe do erro HTTP: {http_err}")
    except requests.exceptions.ConnectionError:
        print("Erro de conexão: Não foi possível alcançar o 'Caminho_backend'. Verifique a URL ou sua rede.")
    except Exception as err:
        print(f"Ocorreu um erro inesperado: {err}")

# Chamando a função
if __name__ == "__main__":
    token = obter_sas_token()
```
