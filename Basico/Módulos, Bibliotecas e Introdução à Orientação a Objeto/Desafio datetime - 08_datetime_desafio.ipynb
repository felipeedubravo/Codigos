## Calculando a idade

Um usuário fornece sua data de nascimento no formato "dd/mm/aaaa". Crie um script Python que calcula a idade do usuário.


```python
from datetime import datetime

data_nascimento_usuario  = input("Infome sua data  de nascimnto(dd/mm/aaaa) :")
print(data_nascimento_usuario)
```

```python
print(type(data_nascimento_usuario))
```

```python
data_nascimento_usuario2 = datetime.strptime(data_nascimento_usuario, "%d/%m/%Y")
print(data_nascimento_usuario2)
```

```python
data_ataul = datetime.now()
print(data_ataul)
```

```python
idade = data_ataul.year - data_nascimento_usuario2.year
print(idade)
```

```python
mes_atual = data_ataul.month
dia_atual = data_ataul.day

mes_n = data_nascimento_usuario2.month
dia_n = data_nascimento_usuario2.day
```

```python
if mes_n > mes_atual :
    idade -= 1
elif mes_n == mes_atual and dia_n > dia_atual:
    idade -= 1
```

```python
print(idade)
```
