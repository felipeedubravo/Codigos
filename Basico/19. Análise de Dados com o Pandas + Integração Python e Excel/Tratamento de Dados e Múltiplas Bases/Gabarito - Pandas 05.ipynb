# Comparando, Tratando e Mesclando DataFrames

## Objetivo

Vamos modificar os IDs para os nomes dos produtos, dos clientes e das lojas, para nossas análises ficarem mais intuitivas futuramente. Para isso, vamos criar um data frame com todos os detalhes.

- Usaremos o método merge para isso e, depois se quisermos, podemos pegar apenas as colunas que queremos do dataframe final.

### Criando nossos dataframes

```python
import pandas as pd

#às vezes precisaremos mudar o encoding. Possiveis valores para testar:
#encoding='latin1', encoding='ISO-8859-1', encoding='utf-8' ou então encoding='cp1252'
vendas_df = pd.read_csv(r'Contoso - Vendas - 2017.csv', 
    sep=';'
)

# Tente primeiro com 'utf-16' (o mais comum para exportações da Contoso/Excel com byte 0xff)
produtos_df = pd.read_csv(
    r'Contoso - Cadastro Produtos.csv', 
    sep=';', 
    encoding='cp1252'
)

# Caso os demais arquivos também apresentem erro de codificação, adicione encoding='latin1' ou 'utf-8' neles:
lojas_df = pd.read_csv(
    r'Contoso - Lojas.csv', 
    sep=';', 
    encoding='cp1252'
)

clientes_df = pd.read_csv(
    r'Contoso - Clientes.csv', 
    sep=';', 
    encoding='cp1252'
)


#usaremos o display para ver todos os dataframes
display(vendas_df)
print("-"*150)
print("-"*150)

display(produtos_df)
print("-"*150)
print("-"*150)

display(lojas_df)
print("-"*150)
print("-"*150)

display(clientes_df)
print("-"*150)
print("-"*150)


```

### Vamos tirar as colunas inúteis do clientes_df ou pegar apenas as colunas que quisermos

```python
clientes_df = clientes_df.drop(
    columns=['Unnamed: 7', 'Unnamed: 8', 'Unnamed: 9', 'Unnamed: 10'], 
    errors='ignore'
) #  Com errors='ignore', o código não quebra se você rodar a célula mais de uma vez!

display(clientes_df.head())
```

```python
# Deleta automaticamente qualquer coluna que comece com "Unnamed"
colunas_unnamed = [col for col in clientes_df.columns if 'Unnamed' in col]
clientes_df.drop(columns=colunas_unnamed, inplace=True, errors='ignore')

display(clientes_df.head())
```

```python
produtos_df.columns = produtos_df.columns.str.replace('ÿ', '').str.strip()
```

```python
import pandas as pd

# 1. Limpeza automática do caractere 'ÿ' e espaços em branco nos nomes das colunas
for df in [vendas_df, produtos_df, lojas_df, clientes_df]:
    df.columns = df.columns.str.replace('ÿ', '').str.strip()

# 2. Removendo colunas Unnamed indesejadas (caso existam)
for df in [vendas_df, produtos_df, lojas_df, clientes_df]:
    colunas_unnamed = [col for col in df.columns if 'Unnamed' in col]
    df.drop(columns=colunas_unnamed, inplace=True, errors='ignore')
```

```python
clientes_df = clientes_df[['ID Cliente', 'E-mail']]
produtos_df = produtos_df[['ID Produto', 'Nome do Produto']]
lojas_df = lojas_df[['ID Loja', 'Nome da Loja']]
display(produtos_df)
```

### Agora vamos juntar os dataframes para ter 1 único dataframe com tudo "bonito"

- Obs: O merge precisa das colunas com o mesmo nome para funcionar. Se não tiver, você precisa alterar o nome da coluna com o .rename

```python
#juntando os dataframes
vendas_df = vendas_df.merge(produtos_df, on='ID Produto')
vendas_df = vendas_df.merge(lojas_df, on='ID Loja')
vendas_df = vendas_df.merge(clientes_df, on='ID Cliente')

#exibindo o dataframe final
display(vendas_df)
```

```python
#vamos renomear o e-mail para ficar claro que é do cliente
vendas_df = vendas_df.rename(columns={'E-mail': 'E-mail do Cliente'})
display(vendas_df)
```
