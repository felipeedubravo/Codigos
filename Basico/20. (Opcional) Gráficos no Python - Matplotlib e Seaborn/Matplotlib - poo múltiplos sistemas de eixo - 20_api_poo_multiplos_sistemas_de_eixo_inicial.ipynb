# Trabalhando com eixos



## Gráfico de linhas

```python
import matplotlib.pyplot as plt

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]

fig, ax = plt.subplots()

ax.plot(meses, vendas)
ax.set_xlabel('Mês')
ax.set_ylabel('Vendas (em milhares de reais)')
ax.set_title('Crescimento das Vendas ao Longo do Ano')

plt.show()
```

## Gráfico de barras

```python
import matplotlib.pyplot as plt

produtos = ['Produto A', 'Produto B', 'Produto C', 'Produto D', 'Produto E']
quantidades = [23, 45, 17, 30, 37]

fig, ax = plt.subplots()

ax.bar(produtos, quantidades)
ax.set_xlabel('Produto')
ax.set_ylabel('Quantidade de vendas')
ax.set_title('Quantidade de vendas por produto')

plt.show()
```

## Gráfico de pizza

```python
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']
quantidades_vendidas = [45, 30, 15, 10]
destaque = [0.1, 0, 0, 0]  # destacar a primeira fatia

fig, ax = plt.subplots()

ax.pie(quantidades_vendidas, labels=categorias, explode=destaque, autopct='%.1f')
ax.set_title('Categorias de produtos vendidos')

plt.show()
```

## Gráfico de barras empilhadas

```python
anos = [2018, 2019, 2020, 2021]
vendas_online = [5, 10, 15, 20]
vendas_loja = [10, 8, 6, 4]

fig, ax = plt.subplots()

ax.bar(anos, vendas_online, label='Online')
ax.bar(anos, vendas_loja, bottom=vendas_online, label='Loja')

ax.set_xlabel('Ano')
ax.set_ylabel('Vendas (em milhões de reais)')
ax.set_title('Vendas online vs loja')

ax.set_xticks(anos)

ax.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

mosasico = 'AB;CD'
fig = plt.figure(figsize=(12,10))
axs = fig.subplot_mosaic(mosasico)


# gráfico 1 - linha

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]

axs['A'].plot(meses, vendas)
axs['A'].set_xlabel('Mês')
axs['A'].set_ylabel('Vendas (em milhares de reais)')
axs['A'].set_title('Crescimento das Vendas ao Longo do Ano')

# gráfico 2 - barras
produtos = ['Produto A', 'Produto B', 'Produto C', 'Produto D', 'Produto E']
quantidades = [23, 45, 17, 30, 37]

axs['B'].bar(produtos, quantidades)
axs['B'].set_xlabel('Produto')
axs['B'].set_ylabel('Quantidade de vendas')
axs['B'].set_title('Quantidade de vendas por produto')

# gráfico 3 - pizza
categorias = ['Eletrônicos', 'Roupas', 'Alimentos', 'Livros']
quantidades_vendidas = [45, 30, 15, 10]
destaque = [0.1, 0, 0, 0]  # destacar a primeira fatia

axs['C'].pie(quantidades_vendidas, labels=categorias, explode=destaque, autopct='%.1f')
axs['C'].set_title('Categorias de produtos vendidos')

# gráfico 4 - baras empilhadas
anos = [2018, 2019, 2020, 2021]
vendas_online = [5, 10, 15, 20]
vendas_loja = [10, 8, 6, 4]

axs['D'].bar(anos, vendas_online, label='Online')
axs['D'].bar(anos, vendas_loja, bottom=vendas_online, label='Loja')
axs['D'].set_xlabel('Ano')
axs['D'].set_ylabel('Vendas (em milhões de reais)')
axs['D'].set_title('Vendas online vs loja')
axs['D'].set_xticks(anos)
axs['D'].legend()

# ajustes finais

fig.suptitle("Evolução de Venda", fontsize=16)
```
