## Gráfico de linha para apresentação


```python
import matplotlib.pyplot as plt
from cycler import cycler

meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]

cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

#plt.plot(meses, vendas, marker='o', markersize=10, linestyle='--', color='green', linewidth=2)
plt.plot(meses, vendas, marker='o', markersize=10, linestyle='--', linewidth=2)

plt.xlabel('Mês')
plt.ylabel('Vendas (em milhares)')
plt.title('Crescimento das Vendas ao Longo do Ano')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from cycler import cycler

# Infomação e dados 
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]
marcado = np.where(vendas == np.min(vendas),'x','')

# paleta de cores
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

# gráfico base
plt. plot(meses, vendas)

# marcadore e anotaç~ões
for posicao, marcador in enumerate(marcado):
    if marcador == 'x':
        cor='red'
        tamalho = 80
    else:
        cor='blue'
        tamalho = 0
    plt.scatter(meses[posicao], vendas[posicao], marker=marcador, s= tamalho, color=cor)

plt.annotate('Menor valor', color='red', xy=(2,78), xytext=(2.2,77.5))

# destaque de região
plt.axvspan(6,11, alpha=0.2, color=cores[1])

plt.text(7, 80, "  Recuperação \nde crescimento" , color='green', fontsize=12, weight='bold')

plt.xlabel('Mês')
plt.ylabel('Vendas (em milhares)')
plt.title('Crescimento das Vendas ao Longo do Ano')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from cycler import cycler

# Infomação e dados 
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
vendas = [85, 90, 78, 92, 88, 96, 94, 98, 100, 102, 105, 108]
marcado = np.where(vendas == np.min(vendas),'x','')

# paleta de cores
cores = plt.get_cmap('Pastel2').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

# gráfico base
plt. plot(meses, vendas)

# marcadore e anotaç~ões
for posicao, marcador in enumerate(marcado):
    if marcador == 'x':
        cor='red'
        tamalho = 80
    else:
        cor='blue'
        tamalho = 0
    plt.scatter(meses[posicao], vendas[posicao], marker=marcador, s= tamalho, color=cor)
    plt.annotate(vendas[posicao], color=cor, xy=(meses[posicao], vendas[posicao]), xytext=(-5,5), textcoords='offset points', fontsize=12, weight='bold')


#plt.annotate('Menor valor', color='red', xy=(2,78), xytext=(2.2,77.5))

# destaque de região
plt.axvspan(6,11, alpha=0.2, color=cores[1])

plt.text(7, 80, "  Recuperação \nde Crescimento" , color='green', fontsize=12, weight='bold')

plt.box(False)
plt.tick_params(axis='x', length=0)
#plt.tick_params(axis='y', length=0)
plt.yticks([])

plt.ylabel('Vendas (em milhares)')
plt.title('Crescimento das Vendas ao Longo do Ano')
#plt.grid()
plt.show()
```
