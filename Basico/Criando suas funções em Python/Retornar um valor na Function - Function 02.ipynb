# Retornar um valor na Function Python

### Estrutura Básica

- Exemplo: vamos criar uma função de cadastro de um Produto. Essa função deve garantir que o produto cadastrado está em letra minúscula.

```python
def cadastrar_produto():
    produto1 = input('Digite o nome do produto que deseja cadastrar')
    produto2 = produto1.casefold() # Cassefld vai deixa as letra minuscula
    produto3 = produto2.strip() # Revome o espaça no inicio e no final da palavra
    return produto3 # return vai mostra o resutado do def "produto3"
```

```python
produto = cadastrar_produto()  # Chamado a nossa função 
print(produto)
```
