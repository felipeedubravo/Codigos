# Biblioteca time

### Estrutura: Uma das bibliotecas mais usadas no Python para trabalhar com datas, fazer o programa esperar algum tempo

```python
import time
```

- Marco Zero: EPOCH (01 de janeiro de 1970 às 00:00:00)
- time() retorna quantos segundos se passaram desde a EPOCH

```python
print(time.time())
```

Isso já pode ser muito útil para medir o tempo que um código leva para rodar

```python
data_inicial = time.time()
# rodo o coigo que eu quero saber quanto tempo demora
for i in range(100000000):
    pass

# page o tempo depois do codigo ter rodado
data_final = time.time()

print(f"meu codigo demorou {data_final - data_inicial} segundos ")
```

- ctime() retorna a data em string

```python
data_hoje = time.ctime()
print(data_hoje)
```

Fazer o programa esperar alguns segundos (muito útil quando temos que esperar um programa ou uma página carregar)

```python

time.sleep(5) # tempo em segundos
```

Pegar informações de dias, hora, minuto, segundo, tudo detalhado:
- gmtime()
- localtime()

```python
data_global = time.gmtime()
data_local = time.localtime()

print(data_global)
print(data_local)
```

Qualquer struct_time permite você pegar as informações da data fazendo:
- gmtime().parametro
- localtime().parametro

```python
dia_hoje = data_local.tm_mday
mes_hoje = data_local.tm_mon
ano_hoje = data_local.tm_year

print("Hoje é dia", dia_hoje)
print("Estamos no mês", mes_hoje)
print("Ano atual",ano_hoje)

print(f"Data Atual: {dia_hoje}/{mes_hoje}/{ano_hoje}")
print("Data Atual: {}/{}/{}".format(dia_hoje,mes_hoje,ano_hoje))
```
