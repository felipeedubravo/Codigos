# Adicionar e Remover itens de uma lista

Adicionar:<br>
lista.append(item)

Remover:<br>
item_removido = lista.pop(indice)<br>
lista.remove(item)

Digamos que você está construindo o controle de produtos da Apple.<br>
E a Apple lançou o IPhone 11 e irá tirar dos seus estoques o IPhone X

```python
produtos = ['apple tv', 'mac', 'iphone x', 'IPad', 'apple watch', 'mac book', 'airpods']

# para trocar o produto
#produtos[2] = "IPhone 11"
print(produtos)

# adicionar ("iphine 11")
produtos.append("iphine 11")
print(produtos)

# remover o ("iphine 11")
produtos.remove("iphine 11")
print(produtos)

item_removido = produtos.pop(2)
print(produtos)
print("Removemos o {} da lista".format(item_removido))
```

```python
produtos2 = ['apple tv', 'mac', 'iphone x', 'IPad', 'apple watch', 'mac book', 'airpods']
print(produtos2)

produtos_remover = 'iphonex'
if produtos_remover in produtos2:
    produtos2.remove('iphonex')
else:
    print("{} não exiteste na lista de produtos".format(produtos_remover))
```
